```python
# from langchain.llms import HuggingFacePipeline
from langchain_huggingface import HuggingFacePipeline
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline
```

```python
# Load the tokenizer and model
tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-large")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-large")
```

```python
# Set up the pipeline
hf_pipeline = pipeline("text2text-generation", model=model, tokenizer=tokenizer, device='cpu')

# Wrap the pipeline in LangChain
llm = HuggingFacePipeline(pipeline=hf_pipeline)
```

```python
from langchain.prompts import PromptTemplate

# Define the prompt template
prompt_template = PromptTemplate(
    input_variables=["job_title"],
    template=(
        "Classify the following job title into its respective domain. "
        "You must choose one from the domains below: \n"
        " - Android Development\n" 
        " - Data Analytics\n"
        " - Design\n"
        " - Digital Marketing\n"
        " - IT & Networking\n" 
        " - Project Management\n"
        " - Software Development\n"
        " - Testing & Support.\n"
        " - Web Development\n\n"
        "Follow the examples below: \n"
        "1. Job Title: Backend -> Domain: Web Development.\n"
        "2. Job Title: Frontend Developers -> Domain: Web Development.\n"
        "3. Job Title: Full Stack Developers -> Domain: Web Development.\n"
        "4. Job Title: Product Manager -> Domain: Project Management.\n"
        "5. Job Title: Reactjs Developers -> Domain: Web Development.\n\n"
        "Job Title: {job_title}\n"
        "Domain:"
    )
)

print(prompt_template.template)
```

```python
from langchain.chains import LLMChain

## Create the chain
# classification_chain = LLMChain(llm=llm, prompt=prompt_template)
classification_chain = prompt_template | llm
```

```python
job_title = "Full Stack Developer - Laravel/MySQL"
response = classification_chain.invoke({"job_title":job_title})
print(f"Domain for '{job_title}': {response}")
```
